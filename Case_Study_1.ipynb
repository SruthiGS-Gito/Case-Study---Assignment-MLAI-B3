```python
# Bank Account Manager

initial_depo = 0 # Initially the amount is set as zero
# Balance is going to affect all functions, hence passing as an arguement
def deposit(balance):
    depo = float(input("Enter amount to deposit : "))
    balance = balance + depo
    print(f"Deposited amount : {depo}\n")
    return balance

def withdraw(balance):
    draw = float(input("Enter amount to withdraw : "))
    if draw > balance:    # If the user is trying to withdraw more than the balance
        print("Insufficient Amount to Withdraw \n")
        return balance

    balance = balance - draw
    print(f"Withdrawn amount : {draw}")
    print(f"Hence after withdrawal of {draw} Current balance : {balance}\n")
    return balance

def balance_enquiry(balance):
    print(f"Balance : {balance}\n")
    return balance

while True:
  op = int(input("Enter the Operation to perform (1/2/3): \n 1) Deposit \n 2) Withdrawal \n 3) Balance Enquiry\n 4) Exit\n"))

  if op == 1:
    initial_depo = deposit(initial_depo)
  elif op == 2:
    initial_depo = withdraw(initial_depo)
  elif op == 3:
    balance_enquiry(initial_depo)
  elif op == 4:
    print("You've exited from doing any operations\n")
    break
  else:
    print("Invalid option, Enter either 1, 2, 3 or 4\n")
```

```python
# Weather Data Analyzer

temperature = float(input("Enter temperature : "))
humidity = float(input("Enter humidity : "))

def weather_condt(tempt, humid):
# Setting Rough figures as threshold

  if tempt > 37 and humid < 45:
    condition = "Sunny and Dry"

  elif tempt > 28 and humid >= 75:
    condition = "Hot and Humid"

  elif humid > 82:
    condition = "Rainy"

  elif tempt < 15:
    condition = "cold"

  else:
    condition = "Not Specific : Most probably a breeze"

  return condition

# Average calculation set as function
def avg(tempt, humid):
    return (tempt + humid) / 2


def main(tempt, humid):
  print(f"According to temperature {tempt} and humidity {humid}, current eather Condition : {weather_condt(tempt, humid)}")
  print(f"Average (Temperature and Humidity) : {avg(tempt, humid)}")

main(temperature, humidity)   # Function call
```
